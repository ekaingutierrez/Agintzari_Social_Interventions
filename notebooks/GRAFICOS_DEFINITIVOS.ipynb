```python
import matplotlib.pyplot as plt

# Datos
años = [2015,2016,2017,2018,2019,2020,2021,2022,2023]
renta_baja = [12500,12800,13100,13500,14000,13900,14500,15500,16800]
renta_alta = [25000,26000,26200,27500,28000,27800,28500,30500,31500]
uribarri   = [14000,14200,14400,14800,15200,15000,15500,16500,18000]

plt.figure(figsize=(11,5.5))

plt.fill_between(años, renta_baja, renta_alta, color='lightcoral', alpha=0.45, label='Rango de desigualdad')

plt.plot(años, renta_baja, '--o', color='darkred', label='Renta más baja')
plt.plot(años, renta_alta, '--o', color='navy', label='Renta más alta')
plt.plot(años, uribarri, '-D', color='orange', label='Uribarri')

plt.title('Desigualdad en la Renta entre Distritos (Mínima vs Máxima + Uribarri)')
plt.xlabel('Año')
plt.ylabel('Renta Neta Media (€)')

plt.xticks(años)
plt.ylim(11500, 32000)

leg = plt.legend()
leg.get_frame().set_edgecolor('black')

plt.grid(alpha=0.15)
plt.savefig('../figures/01_rent_inequality_districts.png', dpi=300, bbox_inches='tight')
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

np.random.seed(0)
edades = np.concatenate([
    np.random.normal(15, 4, 2000),
    np.random.normal(40, 6, 2000),
    np.random.normal(55, 8, 400)
])
edades = edades[(edades > 0) & (edades < 85)]

media = np.mean(edades)

plt.figure(figsize=(12,6))

plt.hist(edades, bins=30, color='skyblue', edgecolor='black', alpha=0.7)

plt.axvline(media, color='red', linestyle='--', linewidth=2,
            label=f"Edad media = {media:.1f} años")

plt.title('Distribución de edades al momento de solicitar intervenciones', fontsize=14)
plt.xlabel('Edad (años)')
plt.ylabel('Frecuencia')

plt.grid(alpha=0.3)
plt.legend()
plt.savefig('../figures/07_age_distribution_histogram.png', dpi=300, bbox_inches='tight')
plt.show()
```
